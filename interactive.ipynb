# SAM3 Interactive Chip Counting

Load model once, change thresholds freely, see results instantly.

**Run from `/tmp/`** to avoid namespace shadow:
```bash
cp /home/ubuntu/sam3_engine.py /tmp/
cp /home/ubuntu/interactive.ipynb /tmp/
cd /tmp && jupyter notebook
```

````python
import os
os.environ["PYTORCH_ALLOC_CONF"] = "expandable_segments:True"

from sam3_engine import SAM3Engine

VIDEO = "/home/ubuntu/dealer.mp4"
engine = SAM3Engine(video_path=VIDEO)
````

````python
# Check current thresholds
engine.get_thresholds()
````

````python
# Change thresholds and run
engine.set_thresholds(
    score_threshold_detection=0.3,
    new_det_thresh=0.7,
)

result = engine.run(
    prompt="stack of poker chips",
    out_dir="/home/ubuntu/sam3-runs/interactive/run1",
    tag="baseline",
)
````

````python
# Try more aggressive detection
engine.set_thresholds(
    score_threshold_detection=0.2,
    new_det_thresh=0.5,
)

result2 = engine.run(
    prompt="stack of poker chips",
    out_dir="/home/ubuntu/sam3-runs/interactive/run2",
    tag="aggressive",
)
````

````python
# Try higher NMS to separate overlapping stacks
engine.set_thresholds(
    score_threshold_detection=0.3,
    new_det_thresh=0.7,
    det_nms_thresh=0.3,
)

result3 = engine.run(
    prompt="stack of poker chips",
    out_dir="/home/ubuntu/sam3-runs/interactive/run3",
    tag="high_nms",
)
````

````python
# Compare results
for r in [result, result2, result3]:
    print(f"unique={r['num_unique_objs']} avg={r['avg_per_frame']} "
          f"max={r['max_per_frame']} elapsed={r['elapsed_sec']}s")
````

````python
# Quick run without overlay (faster, for threshold exploration)
engine.set_thresholds(score_threshold_detection=0.15, new_det_thresh=0.4)
result_quick = engine.run(prompt="stack of poker chips")  # no out_dir = no overlay
````
